**Тема:** «Полиморфизм и Generic классы»

### Описание проекта
Программа демонстрирует принципы полиморфизма, перегрузки и перекрытия методов, а также использование обобщенных типов (Generic) и интерфейсов на примере сборки компьютера из комплектующих.

### Структура проекта:
*   **`IInstallable`** — интерфейс, определяющий метод `Install()` для установки компонента в компьютер.
*   **`Component`** — базовый абстрактный класс, содержащий общие свойства (`Name`, `Price`), виртуальный метод `GetInfo()`, его перегрузку и переопределенный метод `ToString()`.
*   **`CPU`, `RAM`, `GPU`, `HDD`** — производные классы компонентов, которые наследуются от `Component` и реализуют интерфейс `IInstallable`. В каждом классе перекрыт метод `GetInfo()` и перегружены операторы сравнения (`==` и `!=`).
*   **`Computer`** — класс компьютера, хранящий список компонентов и использующий Generic-методы `AddComponent<T>` и `RemoveComponent<T>` с ограничениями типов (`where T`), а также метод расчета общей стоимости `GetTotalPrice()`.
*   **`ComputerFactory`** — статический класс для создания конфигураций компьютера с использованием Generic-методов.

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

// --- ПРОВЕРКА РАБОТЫ ПРОГРАММЫ ---

CPU cpu1 = new CPU("Intel Core i7", 30000, 8);
CPU cpu2 = new CPU("Intel Core i7", 30000, 8);
RAM ram = new RAM("Kingston", 8000, 16);
GPU gpu = new GPU("RTX 4060", 40000, 8);
HDD hdd = new HDD("Seagate", 5000, 1000);

Console.WriteLine(cpu1.ToString());
Console.WriteLine($"Процессоры одинаковые: {cpu1 == cpu2}\n");

// создаем ПК
Computer myPC = ComputerFactory.CreatePC(cpu1);
myPC.AddComponent(ram);
myPC.AddComponent(gpu);
myPC.AddComponent(hdd);

Console.WriteLine("\n--- Состав ПК ---");
myPC.ShowAll();

// проверяем перегрузку метода GetInfo с параметром
gpu.GetInfo("Игровая сборка");

// считаем цену и удаляем деталь
Console.WriteLine($"\nОбщая цена: {myPC.GetTotalPrice()} руб.");
myPC.RemoveComponent(hdd);
Console.WriteLine($"Цена без диска: {myPC.GetTotalPrice()} руб.");


// --- КЛАССЫ И ИНТЕРФЕЙСЫ ---

// интерфейс для установки деталей
public interface IInstallable
{
    void Install();
}

// базовый абстрактный класс
public abstract class Component
{
    public string Name { get; set; }
    public double Price { get; set; }

    // конструктор
    public Component(string name, double price)
    {
        Name = name;
        Price = price;
    }

    // виртуальный метод 
    public virtual void GetInfo()
    {
        Console.WriteLine($"Деталь: {Name}, Цена: {Price}");
    }

    // перегрузка  
    public void GetInfo(string note)
    {
        Console.WriteLine($"Деталь: {Name} ({note}), Цена: {Price}");
    }

    public override string ToString()
    {
        return $"{Name} - {Price} руб.";
    }
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string name, double price, int cores) : base(name, price)
    {
        Cores = cores;
    }

    // перекрытие 
    public override void GetInfo()
    {
        Console.WriteLine($"Процессор: {Name}, Ядер: {Cores}, Цена: {Price}");
    }

    // реализация метода из интерфейса IInstallable
    public void Install()
    {
        Console.WriteLine($"Процессор {Name} установлен.");
    }

    // перегрузка операторов сравнения == и !=
    public static bool operator ==(CPU a, CPU b)
    {
        return a.Name == b.Name && a.Cores == b.Cores;
    }
    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }
}

public class RAM : Component, IInstallable
{
    public int Memory { get; set; }

    public RAM(string name, double price, int memory) : base(name, price)
    {
        Memory = memory;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Память: {Name}, Объем: {Memory} ГБ, Цена: {Price}");
    }

    public void Install()
    {
        Console.WriteLine($"Память {Name} установлена.");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        return a.Name == b.Name && a.Memory == b.Memory;
    }
    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }
}

public class GPU : Component, IInstallable
{
    public int VideoMemory { get; set; }

    public GPU(string name, double price, int videoMemory) : base(name, price)
    {
        VideoMemory = videoMemory;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Видеокарта: {Name}, Память: {VideoMemory} ГБ, Цена: {Price}");
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта {Name} установлена.");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        return a.Name == b.Name && a.VideoMemory == b.VideoMemory;
    }
    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }
}

public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }

    public HDD(string name, double price, int capacity) : base(name, price)
    {
        Capacity = capacity;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Диск: {Name}, Объем: {Capacity} ГБ, Цена: {Price}");
    }

    public void Install()
    {
        Console.WriteLine($"Диск {Name} установлен.");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        return a.Name == b.Name && a.Capacity == b.Capacity;
    }
    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }
}

public class Computer
{
    // список всех деталей внутри компьютера
    private List<Component> _components = new List<Component>();

    // generic-метод добавления 
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        component.Install();
        _components.Add(component);
    }

    // generic-метод удаления
    public void RemoveComponent<T>(T component) where T : Component
    {
        _components.Remove(component);
        Console.WriteLine($"Деталь {component.Name} удалена.");
    }

    // подсчет общей стоимости через цикл
    public double GetTotalPrice()
    {
        double sum = 0;
        foreach (var item in _components)
        {
            sum += item.Price;
        }
        return sum;
    }

    // полиморфизм
    public void ShowAll()
    {
        foreach (var item in _components)
        {
            item.GetInfo();
        }
    }
}
public static class ComputerFactory
{
    public static Computer CreatePC<T>(T startComponent) where T : Component, IInstallable
    {
        Computer pc = new Computer();
        pc.AddComponent(startComponent);
        return pc;
    }
}
```
