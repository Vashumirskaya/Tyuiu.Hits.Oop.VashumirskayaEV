**Тема:** «Наследование, Абстрактный класс и интерфейсы»

**Цель задания:** создать классы, представляющие структуру коммерческого предприятия, используя простое, сложное и множественное наследование.

### Описание задачи:
Создать иерархию классов сотрудников компании с демонстрацией различных видов наследования.

#### 1. Базовый класс (Employee)
*   **Поля:** `Name` (имя сотрудника), `Position` (должность), `Salary` (зарплата).
*   **Методы:** `PrintInfo()` (вывод информации о сотруднике).

#### 2. Простое наследование (Manager)
*   Наследуется от `Employee`.
*   **Поля:** `Department` (отдел).
*   Переопределен метод `PrintInfo()` для вывода отдела.

#### 3. Сложное наследование (ProjectManager)
*   Наследуется от `Manager` (цепочка: Employee -> Manager -> ProjectManager).
*   **Поля:** `ProjectName` (название проекта).
*   Переопределен метод `PrintInfo()` для вывода проекта.

#### 4. Множественное наследование (через интерфейсы)
Реализация множественного наследования для `ProjectManager` с помощью интерфейсов:
*   `IReportable`: метод `GenerateReport()` (возвращает строку с отчетом).
*   `ITeamLeader`: метод `GetTeamMembers()` (возвращает список/массив членов команды).

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
// --- ТЕСТИРОВАНИЕ ---

Employee emp = new Employee("Иван Иванов", "Программист", 80000f);
emp.PrintInfo();

Console.WriteLine();

Manager mgr = new Manager("Анна Смирнова", "Руководитель отдела", 120000f, "IT-отдел");
mgr.PrintInfo();

Console.WriteLine();

ProjectManager pm = new ProjectManager("Елизавета Вашумирская", "Проектный менеджер", 150000f, "IT-отдел", "Разработка нового приложения");
pm.PrintInfo();

Console.WriteLine();
Console.WriteLine(pm.GenerateReport());

Console.WriteLine("Команда проекта:");
foreach (string member in pm.GetTeamMembers())
{
    Console.WriteLine($"- {member}");
}


// --- СОЗДАНИЕ БАЗОВОГО КЛАССА ---
public class Employee
{
    // Поля
    public string Name { get; set; }
    public string Position { get; set; }
    public float Salary { get; set; }

    // Конструктор
    public Employee(string name, string position, float salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    // Метод
    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, Должность: {Position}, Зарплата: {Salary} руб.");
    }
}

// --- ПРОСТОЕ НАСЛЕДОВАНИЕ ---
public class Manager : Employee
{
    // Поле
    public string Department { get; set; }

    // Конструктор
    public Manager(string name, string position, float salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    // Переопределенный метод
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

// --- МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (ИНТЕРФЕЙСЫ) ---
public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers(); 
}

// --- СЛОЖНОЕ НАСЛЕДОВАНИЕ ---
// Наследует Manager (сложное наследование) и реализует интерфейсы (множественное)
public class ProjectManager : Manager, IReportable, ITeamLeader
{
    // Поле
    public string ProjectName { get; set; }

    // Конструктор
    public ProjectManager(string name, string position, float salary, string department, string projectName)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
    }

    // Переопределенный метод
    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
    }

    // Реализация методов из интерфейсов
    public string GenerateReport()
    {
        return $"[ОТЧЕТ] Проект «{ProjectName}» выполняется по графику.";
    }

    public string[] GetTeamMembers()
    {
        return new string[] { "Арина (Аналитик)", "Иван (Разработчик)", "Петр (Тестировщик)" };
    }
}
```
