# Индивидуальный проект

**Название проекта:** Система управления кредитными картами с использованием принципов ООП


**Описание проекта:**
Создать базовый класс `Movie` в C#, который будет представлять информацию о фильмах. На основе этого класса разработать несколько производных классов, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

### Требования к базовому классу `Movie`:

*   **Атрибуты:** Название (`Title`), Режиссер (`Director`), Год выпуска (`ReleaseYear`), Продолжительность (`Duration`), Возрастной рейтинг (`AgeRating`), Кассовые сборы (`BoxOffice`).
*   **Методы:**
    *   `GetInfo()`: метод для получения информации о фильме в виде строки.
    *   `Watch()`: метод для вывода сообщения о просмотре фильма.
    *   `Rate()`: метод для оценки фильма.
    *   `PlayTrailer()`: метод запуска трейлера фильма.
    *   `BuyTicket()`: метод для покупки билета на сеанс.

### Требования к производным классам:

1.  **Документальный фильм (`Documentary`):** Демонстрирует **простое наследование**. Должен содержать дополнительные атрибуты, такие как Тематика (`Theme`). Метод `Watch()` должен быть переопределен для добавления информации о тематике при просмотре документального фильма.
2.  **Игровой фильм (`FeatureFilm`) и Научная фантастика (`SciFiFilm`):** Демонстрируют **сложное наследование**. Класс `FeatureFilm` добавляет атрибут Жанр (`Genre`) и переопределяет метод оценки. Класс `SciFiFilm` наследуется от `FeatureFilm`, углубляя иерархию, и добавляет атрибут Главная концепция (`CoreConcept`).
3.  **Анимационный фильм (`AnimatedMovie`):** Демонстрирует **множественное наследование** (через наследование базового класса и реализацию интерфейсов `IMerchandisable` и `IAwardWinning`). Должен содержать дополнительные атрибуты, такие как Студия анимации (`AnimationStudio`).
    *   Метод `GetInfo()` должен быть переопределен для включения информации о студии анимации в описании фильма.

### Дополнительное задание:

*   Добавьте к существующим классам конструкторы классов с использованием геттеров и сеттеров.
*   Добавьте 3-4 новых атрибута и метода.
*   Реализуйте взаимодействие объектов между собой.
*   Продемонстрируйте виды наследования: простое, сложное и множественное.

```
// --- ДЕЙСТВИЯ (ОБЪЕКТЫ) ---

// 1. Простое наследование (Movie -> Documentary)
Documentary doc = new Documentary("Планета Земля", "Аластер Фотергилл", 2006, "Природа", 530, "0+");
Console.WriteLine(doc.GetInfo());
doc.PlayTrailer();
doc.Watch();
doc.Rate();

Console.WriteLine();

// 2. Сложное наследование (Movie -> FeatureFilm -> SciFiFilm)
SciFiFilm feature = new SciFiFilm("Начало", "Кристофер Нолан", 2010, "Научная фантастика", 148, "12+", "Осознанные сновидения");
Console.WriteLine(feature.GetInfo());
feature.BuyTicket();
feature.Watch();
feature.Rate();

Console.WriteLine();

// 3. Множественное наследование (AnimalMovie наследует Movie и реализует IMerchandisable, IAwardWinning)
AnimatedMovie anim = new AnimatedMovie("Унесенные призраками", "Хаяо Миядзаки", 2001, "Studio Ghibli", 125, "12+");
Console.WriteLine(anim.GetInfo());
anim.Watch();
anim.Rate();
// Методы из интерфейсов:
anim.SellMerch();
anim.ReceiveAward("Оскар");

Console.WriteLine("\n--- ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ ---");
feature.CompareReleaseYear(anim);
anim.Recommend(doc);


// --- КЛАССЫ И ИНТЕРФЕЙСЫ ---

// ИНТЕРФЕЙСЫ (для множественного наследования)
public interface IMerchandisable { void SellMerch(); }
public interface IAwardWinning { void ReceiveAward(string awardName); }

// БАЗОВЫЙ КЛАСС MOVIE
public class Movie
{
    private string _title;
    private string _director;
    private int _releaseYear;
    
    // 3 новых атрибута
    public int Duration { get; set; } 
    public string AgeRating { get; set; }
    public decimal BoxOffice { get; set; } = 0;

    public string Title { get { return _title; } set { _title = string.IsNullOrEmpty(value) ? "Без названия" : value; } }
    public string Director { get { return _director; } set { _director = string.IsNullOrEmpty(value) ? "Неизвестный режиссер" : value; } }
    public int ReleaseYear { get { return _releaseYear; } set { _releaseYear = (value >= 1895) ? value : 1895; } }

    public Movie(string title, string director, int releaseYear, int duration, string ageRating)
    {
        Title = title;
        Director = director;
        ReleaseYear = releaseYear;
        Duration = duration;
        AgeRating = ageRating;
    }

    // Методы взаимодействия
    public void CompareReleaseYear(Movie otherMovie)
    {
        if (this.ReleaseYear > otherMovie.ReleaseYear) Console.WriteLine($"Фильм «{this.Title}» вышел позже, чем «{otherMovie.Title}».");
        else if (this.ReleaseYear < otherMovie.ReleaseYear) Console.WriteLine($"Фильм «{this.Title}» вышел раньше, чем «{otherMovie.Title}».");
        else Console.WriteLine($"Фильмы «{this.Title}» и «{otherMovie.Title}» вышли в один год.");
    }

    public void Recommend(Movie otherMovie)
    {
        Console.WriteLine($"Если вам понравился «{this.Title}», рекомендуем также посмотреть «{otherMovie.Title}».");
    }

    // Базовые методы
    public virtual string GetInfo()
    {
        return $"Фильм: {Title}, Режиссер: {Director}, Год: {ReleaseYear}, Длительность: {Duration} мин, Рейтинг: {AgeRating}";
    }

    public virtual void Watch()
    {
        Console.WriteLine($"Вы смотрите фильм {Title}.");
    }

    public virtual void Rate()
    {
        Console.WriteLine($"Вы оценили фильм {Title} на 8 из 10.");
    }

    // 2 новых метода
    public void PlayTrailer()
    {
        Console.WriteLine($"[Плеер] Запущен трейлер фильма: {Title}");
    }

    public void BuyTicket()
    {
        Console.WriteLine($"Билет на сеанс «{Title}» куплен!");
    }
}

// 1. ПРОСТОЕ НАСЛЕДОВАНИЕ
public class Documentary : Movie
{
    private string _theme;
    public string Theme { get { return _theme; } set { _theme = value; } }

    public Documentary(string title, string director, int releaseYear, string theme, int duration, string ageRating)
        : base(title, director, releaseYear, duration, ageRating)
    {
        Theme = theme;
    }

    public override void Watch()
    {
        Console.WriteLine($"Вы смотрите документальный фильм {Title} на тему: {Theme}.");
    }
}

// 2. СЛОЖНОЕ НАСЛЕДОВАНИЕ (Movie -> FeatureFilm -> SciFiFilm)
// Сначала Игровой фильм
public class FeatureFilm : Movie
{
    private string _genre;
    public string Genre { get { return _genre; } set { _genre = value; } }

    public FeatureFilm(string title, string director, int releaseYear, string genre, int duration, string ageRating)
        : base(title, director, releaseYear, duration, ageRating)
    {
        Genre = genre;
    }

    public override void Rate()
    {
        Console.WriteLine($"Для жанра {Genre} фильм {Title} получает особую оценку: 10 из 10!");
    }
}

// Затем Научная фантастика (наследует FeatureFilm)
public class SciFiFilm : FeatureFilm
{
    public string CoreConcept { get; set; }

    public SciFiFilm(string title, string director, int releaseYear, string genre, int duration, string ageRating, string coreConcept)
        : base(title, director, releaseYear, genre, duration, ageRating)
    {
        CoreConcept = coreConcept;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Главная концепция: {CoreConcept}";
    }
}

// 3. МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (Наследует класс Movie + Реализует 2 интерфейса)
public class AnimatedMovie : Movie, IMerchandisable, IAwardWinning
{
    private string _animationStudio;
    public string AnimationStudio { get { return _animationStudio; } set { _animationStudio = value; } }

    public AnimatedMovie(string title, string director, int releaseYear, string animationStudio, int duration, string ageRating)
        : base(title, director, releaseYear, duration, ageRating)
    {
        AnimationStudio = animationStudio;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Студия анимации: {AnimationStudio}";
    }

    // Реализация методов из интерфейсов
    public void SellMerch()
    {
        Console.WriteLine($"В магазине продаются фигурки персонажей от студии {AnimationStudio} из фильма {Title}.");
    }

    public void ReceiveAward(string awardName)
    {
        Console.WriteLine($"Фильм {Title} получил престижную награду: {awardName}!");
    }
}
```
