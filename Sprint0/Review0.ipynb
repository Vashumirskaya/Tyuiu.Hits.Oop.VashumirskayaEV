### Задача:

Разработать проект для управления растениями в саду, используя основные концепции объектно-ориентированного программирования (ООП) в C#. Вам нужно создать классы, которые будут описывать разные типы растений, используя наследование и полиморфизм.

### Требования к базовому классу Растение (Plant):

Поля (свойства):
- Имя (string): название растения (например, «Роза»).
- Тип (string): тип растения (например, «Цветок» или «Дерево»).
- Высота (float): высота растения в сантиметрах.

Методы:
- Цветение(): метод, который выводит сообщение о том, что растение цветет.
- Растет(float рост): метод для увеличения высоты растения на заданное значение.


### Требования к классу Цветок (Flower) — наследуется от класса Растение:

Поля (свойства):
- Цвет (string): цвет цветка (например, «Красный»).
- КоличествоЛепестков (int): количество лепестков у цветка.

Методы:
- Цветение(): переопределить метод из класса Растение, чтобы вывести специальное сообщение для цветков.
- Увядать(): метод, который сообщает, что цветок начинает увядать.


### Требования к классу Дерево (Tree) — наследуется от класса Растение:

Поля (свойства):
- Возраст (int): возраст дерева в годах.
- ТипЛистьев (string): тип листьев (например, «Хвойные» или «Лиственные»).

Методы:
- Цветение(): переопределить метод из класса Растение с сообщением, специфичным для дерева.
- Вырастить(int годы): метод, который увеличивает возраст дерева и сообщает о новом возрасте.


### Полиморфизм
Каждый класс (цветок и дерево) должен использовать свой вариант метода Цветение(), чтобы продемонстрировать полиморфизм. Например:

- При вызове Цветение() для цветка должно выводиться: "Цветок [Имя] распускается и радует глаз!"
- При вызове Цветение() для дерева должно выводиться: "Дерево [Имя] покрывается новыми листьями!"


### Задание:
1. Создайте объекты для одного цветка и одного дерева.
2. Вызовите методы, чтобы показать, как растения цветут и растут.
3. Реализуйте полиморфизм через переопределение метода Цветение() в классах-наследниках, чтобы отработать концепцию.

```
Flower myFlower = new Flower("Роза", "Цветок", 25.5f, "Красный", 15);
Tree myTree = new Tree("Дуб", "Дерево", 150.0f, 10, "Лиственные");

myFlower.Bloom();
myFlower.Grow(5.0f);
myFlower.Fade();

Console.WriteLine();

myTree.Bloom();
myTree.Grow(20.0f);
myTree.GrowOlder(2);


public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветет.");
    }

    public void Grow(float amount)
    {
        Height += amount;
        Console.WriteLine($"{Type} {Name} выросло на {amount} см. Текущая высота: {Height} см.");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, string type, float height, string color, int petalCount) 
        : base(name, type, height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Fade()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать.");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, string type, float height, int age, string leafType) 
        : base(name, type, height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} стало старше на {years} года/лет. Теперь ему {Age} лет.");
    }
}
```
